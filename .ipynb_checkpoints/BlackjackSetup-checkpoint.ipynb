```python
import gymnasium as gym
from typing import Optional
```

```python
import time
env = gym.make("Blackjack-v1", render_mode = "human")

for i in range(10):
    episode_over = False
    env.reset()
    while not episode_over:
        env.render()
        action = env.action_space.sample()
        observation, reward, terminated, truncated, info = env.step(action)
        time.sleep(1)
        env.render()
        time.sleep(1)
        episode_over = terminated or truncated
env.close()
```

```python
class BlackjackTable(gym.Env):
    def __init__(self):
        self.observation_space = gym.spaces.Dict(
            {"bankroll": gym.spaces.Box(low = 0,high = np.inf, shape = (1,), dtype = np.int64)
                                        }
        )
        self.action_space = gym.spaces.OneOf((gym.spaces.Box(low = 1, high = 1000, shape = (1,), dtype = np.int64), gym.spaces.Discrete(2)))
        self.deck = np.arange(52)
        self.bankroll = 0
        
    def reset(self, seed: Optional[int] = None, options: Optional[dict] = None):
        super().reset(seed=seed)
        self.np_random.shuffle(self.deck)
        self.bankroll = 100

    def _get_obs(self):
        return {"bankroll": self.bankroll, "deck":self.deck}

    def _get_info(self):
        return {"yuh":0}

    def step(self, action):
        return self._get_obs(), 1, False, False, self._get_info()
        
```
